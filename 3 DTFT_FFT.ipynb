```python
import numpy as np

import matplotlib.pyplot as plt
```

```python
n=np.arange(40)
x=0.9**n
```

```python
N_fft=1024

X=np.fft.fft(x,N_fft)
Xshift=np.fft.fftshift(X)
```

```python
w=np.fft.fftfreq(N_fft)*2*np.pi
wshift=np.fft.fftshift(w)
```

```python
mag=np.abs(Xshift)
phase=np.angle(Xshift)
```

```python
plt.figure
plt.subplot(2,1,1)
plt.plot(wshift,mag)

plt.subplot(2,1,2)
plt.plot(wshift,phase)

maxpoint=np.argmax(mag)
wmax=wshift[maxpoint]
print(wmax)

```

```python
import numpy as np
import matplotlib.pyplot as plt

n=np.arange(64)
x=np.sin(0.6*np.pi*n)
N=2048

X=np.fft.fft(x,N)
Xshift=np.fft.fftshift(X)

w=np.fft.fftfreq(N)*2*np.pi
wshift=np.fft.fftshift(w)

mag=np.abs(Xshift)
phase=np.angle(Xshift)

maxpoint=np.argmax(mag)
wmax=wshift[maxpoint]
print(wmax)
plt.figure
plt.subplot(2,1,1)
plt.plot(wshift,mag)

plt.subplot(2,1,2)
plt.plot(wshift,phase)

```

```python
import numpy as np
import matplotlib.pyplot as plt

N=256
X=np.zeros(N)

X[20]=1
X[N-20]=1

x=np.fft.ifft(X)

k=np.arange(N)
n=np.arange(N)

plt.figure
plt.subplot(2,2,1)
plt.plot(k,np.abs(X) )

plt.subplot(2,2,2)
plt.plot(k,np.angle(X) )

plt.subplot(2,2,3)
plt.plot(n,np.real(x) )

plt.subplot(2,2,4)
plt.plot(n,np.imag(x) )
```
