# ============================================================
# Question 2.1
#
# In this problem we study basic continuous-time signal
# transformations on a cosine signal.
#
# We examine:
# • Time shifting
# • Time compression
# • Amplitude inversion
# • Amplitude attenuation
#
# by comparing transformed versions of x(t).
# ============================================================

```python
t = np.linspace(0,4*np.pi,3000)
```

```python
x_t = 2*np.cos(t)
```

Μετασχηματισμοί

```python
x_shift = 2*np.cos(t-np.pi)
x_comp = 2*np.cos(2*t)
x_inv = -0.5*x_t
```

```python

plt.figure(figsize=(10,5))
plt.plot(t,x_t,label='x(t)')
plt.plot(t,x_shift,label='x(t-pi)')
plt.plot(t,x_comp,label='x(2t)')
plt.plot(t,x_inv,label='-0.5x(t)')
plt.legend()
plt.xlabel('t')
plt.ylabel('x(t)')
plt.show()
```

# ============================================================
# Question 2.2
#
# In this problem we study transformations of a
# discrete-time signal.
#
# We investigate:
# • Signal delay
# • Time reversal
# • Amplitude scaling
#
# using stem plots for discrete sequences.
# ============================================================

```python
n = np.arange(-20,20)
```

```python
x_n = 2*n*np.cos((np.pi/4)*n)
```

Μετασχηματισμοί

```python
x_delay =2*n*np.cos((np.pi/4)*(n-3))
x_rev = 2*n*np.cos((np.pi/4)*(-n))
x_amp = 2*n*np.cos((np.pi/4)*n)
```

```python
plt.stem(n,x_n)
plt.xlabel('n')
plt.ylabel('x(n)')
plt.show()
```

```python
plt.stem(n,x_delay)
plt.xlabel('n')
plt.ylabel('x(n)')
plt.show()
```

```python
plt.stem(n,x_rev)
plt.xlabel('n')
plt.ylabel('x(n)')
plt.show()
```

```python
plt.stem(n,x_amp)
plt.xlabel('n')
plt.ylabel('x(n)')
plt.show()
```

# ============================================================
# Question 2.3
#
# In this problem we study uniform quantization of a
# continuous-time signal.
#
# We compare the original and quantized signals,
# compute the quantization error, and verify the
# theoretical quantization error bound.
# ============================================================

```python
t = np.linspace(0,2*np.pi,2000)
```

```python
x_t = 2*np.cos(t)
```

```python
D = 0.25
```

```python
xq = D*np.round(x_t/D)
```

```python
e = xq - x_t
```

```python
plt.plot(t,x_t,label='x(t)')
plt.plot(t,xq,label='xq(t)')
plt.legend()
plt.xlabel('t')
plt.ylabel('x(t)')
plt.show()
```

```python
plt.plot(t,e)
plt.xlabel('t')
plt.ylabel('e(t)')
plt.show()
```

```python
max_error = max(abs(e))
```

```python
print(max_error)
```

```python
sigma = -8
f0 = 60
f1 = 200
a = -0.005
eps = 1e-12
```

```python
f = np.linspace(0,1000,4096)
```

```python
def X_of_f(f,sigma,f0,f1,a):
  alpha = -sigma


  Xf0 = alpha*(1/(alpha**2 +(2*np.pi*(f-f0)**2)) + 1/(alpha**2 +(2*np.pi*(f+f0)**2)))

  Xf1 = alpha*(1/(alpha**2 +(2*np.pi*(f-f1)**2)) + 1/(alpha**2 +(2*np.pi*(f+f1)**2)))

  Xf = Xf0 + a*Xf1
  return Xf

  metro = np.abs(Xf)
  m_max = max(abs(metro))
  m_min = min(abs(metro))

  plt.plot(f,metro)
  plt.show()

  mdb = 20*np.log10((metro/m_max)+eps)
  plt.plot(f,mdb)
  plt.show()

  Ddb = 20*np.log10((m_max/m_min)+eps)
  print(Ddb)

  print(m_max)
  print(m_min)
```

```python
Xf = X_of_f(f,sigma,f0,f1,a)
```

```python
print(Xf)
```

```python
metro = np.abs(Xf)
m_max = np.max(metro)
m_min = np.min(metro)

plt.plot(f,metro)
plt.show()

mdb = 20*np.log10((metro/m_max)+eps)
plt.plot(f,mdb)
plt.show()

Ddb = 20*np.log10((m_max/m_min)+eps)
print(Ddb)

print(m_max)
print(m_min)
```
