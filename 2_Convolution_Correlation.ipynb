```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.linspace(-1,4,4000)
```

```python
Dt = t[2] - t[1]
```

```python
def u(t):
  return np.where(t>=0,1,0)
```

```python
x = u(t) - u(t-2)
h = u(t) - u(t-1)
```

```python
y = Dt*np.convolve(x,h)
```

Ορίζουμε νεο χρόνο για την συνέλιξη και επειδη ειναι διακριτη θα εχει μήκος L+M-1

```python
tconv = np.linspace(-2,8,len(y))
```

```python
plt.plot(t,x,label="x(t)")
plt.plot(t,h,label ="h(t)")
plt.plot(tconv,y,label = "y(t)")
plt.xlabel("t")
plt.ylabel("Σήματα")
plt.legend()
plt.show()
```

```python
x = np.array([1,2,1])
```

```python
h = np.array([1,1,-1])
```

```python
L = len(x)+len(y)-1
```

```python
y = np.zeros(L)
```

```python
for n in range(L):
  for k in range(len(x)):
    if 0<= n-k < len(h):
      y[n] += x[k]*h[n-k]
```
